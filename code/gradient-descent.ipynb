**Gradient Descent on a 1-D function**
$$f(w) = w^4 - w^3 - 12w^2 + 15w + 5$$

```python
import numpy as np
import matplotlib.pyplot as plt

def f(w):
    return w**4 - w**3 - 12*w**2 + 15*w + 5

def df(w):
    return 4*w**3 - 3*w**2 - 24*w + 15

w = np.linspace(-4, 4, 100) 
plt.plot(w, f(w))

w0 = -4
eta = 0.02

plt.plot(w0, f(w0), 'ro')

w = w0

for i in range(10):
    w = w - eta * df(w)  # Update w using gradient descent
    plt.plot(w, f(w), 'bo-')  # Plot the random point)

plt.xlabel('w')
plt.ylabel('f(w)')
plt.title('Function to Optimize')
plt.grid(True)
plt.show()
```

```python
def q(x,y):
    return x*y

def r(x,q):
    return q + x**2

def L(r):
    return r**2

q = np.linspace(-5, 5, 100)
r = np.linspace(-5, 5, 100)

# Initialize the figure workspace
fig = plt.figure()
# Instantiate a 3D axes framework
ax = plt.axes(projection='3d')

ax.plot3D(q, r, L(r), 'red')
x = 2
y = -3
ax.scatter(x, y, L(r), color='blue')
ax.set_title('3D Line Plot Example')

plt.show()
```
